# Modelo principal — P1L2

Notebook de ejecución del modelo estructural principal. Ordena la generación de geometría, validación, análisis OpenSees y exportación para Unity.

## Preparación

```python
from pathlib import Path
import json, subprocess, sys
ROOT=Path.cwd().parent if Path.cwd().name=='notebooks' else Path.cwd()
SCRIPTS=ROOT/'scripts'
OUT=ROOT/'outputs'
print(ROOT)
```

## 1. Generar el modelo manual

```python
r=subprocess.run([sys.executable,str(SCRIPTS/'generar_modelo_manual.py')],cwd=ROOT,text=True,capture_output=True)
print(r.stdout)
if r.returncode: print(r.stderr); raise RuntimeError('Falló la generación')
model=json.loads((OUT/'modelo_3d_manual.json').read_text(encoding='utf-8'))
print('Nodos:',len(model.get('nodes',[])))
print('Elementos:',len(model.get('elements',[])))
print('Losas:',len(model.get('slabs',[])))
```

## 2. Verificar geometría, apoyos y cargas

```python
r=subprocess.run([sys.executable,str(SCRIPTS/'verificar_modelo.py')],cwd=ROOT,text=True,capture_output=True)
print(r.stdout)
if r.returncode: print(r.stderr); raise RuntimeError('Falló la validación')
```

## 3. Análisis estructural OpenSees

```python
r=subprocess.run([sys.executable,str(SCRIPTS/'modelo_opensees_3d.py')],cwd=ROOT,text=True,capture_output=True)
print(r.stdout)
if r.returncode: print(r.stderr); raise RuntimeError('Falló OpenSees')
```

## 4. Exportar datos para Unity

```python
r=subprocess.run([sys.executable,str(SCRIPTS/'crear_visualizador_3d_modelo.py')],cwd=ROOT,text=True,capture_output=True)
print(r.stdout)
if r.returncode: print(r.stderr); raise RuntimeError('Falló la exportación')
print('Los archivos de Unity quedan en:', ROOT/'UnityVisualization')
```
